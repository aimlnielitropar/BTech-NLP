```python
# ============================================================
# PART A: WordNet similarity measures
# ============================================================
from nltk.corpus import wordnet as wn

dog = wn.synset("dog.n.01")
cat = wn.synset("cat.n.01")
car = wn.synset("car.n.01")

pairs = [("dog", dog, "cat", cat), ("dog", dog, "car", car)]
for n1, s1, n2, s2 in pairs:
    print(f"{n1}-{n2}:")
    print(f"  Path similarity:  {s1.path_similarity(s2):.3f}")
    print(f"  Wu-Palmer sim:    {s1.wup_similarity(s2):.3f}")
    print(f"  Lowest common hypernym: {s1.lowest_common_hypernyms(s2)}")
```

```python
# ============================================================
# PART B: FrameNet exploration
# ============================================================
from nltk.corpus import framenet as fn

frames = fn.frames(r"(?i)commerce")
print("Frames matching 'commerce':")
for f in frames[:3]:
    print(" -", f.name, ":", f.definition[:120], "...")

if frames:
    example_frame = frames[0]
    print(f"\nFrame elements (roles) for '{example_frame.name}':")
    for fe_name in list(example_frame.FE.keys())[:6]:
        print(" -", fe_name)

print("\nIndustry link: FrameNet-style semantic roles still power event extraction")
print("from financial news and legal documents (e.g. 'who bought what, from whom, for how much').")
```

```python
# ============================================================
# PART C: Stemmer comparison -- Porter vs Lancaster vs Snowball
# ============================================================
from nltk.stem import PorterStemmer, LancasterStemmer, SnowballStemmer

porter, lancaster, snowball = PorterStemmer(), LancasterStemmer(), SnowballStemmer("english")

words = ["running", "flies", "studies", "national", "generously", "connection"]

header = f"{'WORD':14s}{'PORTER':12s}{'LANCASTER':12s}{'SNOWBALL'}"
print(header)
for w in words:
    print(f"{w:14s}{porter.stem(w):12s}{lancaster.stem(w):12s}{snowball.stem(w)}")
```

```python
# ============================================================
# PART D: A POS tagger trained from scratch on a tagged corpus
# ============================================================
from nltk.corpus import treebank
import nltk

train_data = treebank.tagged_sents()[:3000]
test_data = treebank.tagged_sents()[3000:3200]

unigram_tagger = nltk.UnigramTagger(train_data)
bigram_tagger = nltk.BigramTagger(train_data, backoff=unigram_tagger)

accuracy = bigram_tagger.accuracy(test_data) if hasattr(bigram_tagger, "accuracy") else bigram_tagger.evaluate(test_data)
print(f"Bigram tagger (with unigram backoff) accuracy on held-out Treebank data: {accuracy:.3f}")

sample = "The quick fox jumps over lazy dogs".split()
print("\nTagged sample:", bigram_tagger.tag(sample))
```
