```python
# ============================================================
# Run ONE sentence through morphological -> syntactic -> semantic -> pragmatic analysis
# ============================================================
import nltk

sentence = "The boys who study NLP will explain it clearly."

# --- Morphological level: tokenization + simple stemming ---
tokens = nltk.word_tokenize(sentence)
print("Tokens:", tokens)

def naive_stem(word):
    for suf in ["ing", "ed", "s"]:
        if word.lower().endswith(suf) and len(word) > len(suf) + 2:
            return word[: -len(suf)]
    return word

morph = [(t, naive_stem(t)) for t in tokens]
print("Morphological analysis (token, stem):", morph)

# --- Syntactic level: POS tagging (a lightweight stand-in for a full parse) ---
pos_tags = nltk.pos_tag(tokens)
print("\nSyntactic level (POS tags):", pos_tags)

# --- Semantic level: toy frame extraction using POS patterns ---
# Very simplified: find main verb + its likely arguments
verbs = [w for w, t in pos_tags if t.startswith("VB")]
nouns = [w for w, t in pos_tags if t.startswith("NN") or t == "PRP"]
frame = {"predicate": verbs, "arguments": nouns}
print("\nSemantic level (toy frame):", frame)

# --- Pragmatic level: discourse-context note (rule-based) ---
if "it" in [t.lower() for t in tokens]:
    print("\nPragmatic level: pronoun \"it\" requires discourse context to resolve its referent.")
```

```python
# ============================================================
# Applied example: toy rule-based English -> Hindi translation
# (A real system would use IBM Model 1 / NMT -- see Practical 8)
# ============================================================
bilingual_dict = {
    "the": "", "boys": "ladke", "who": "jo", "study": "padhte hain",
    "nlp": "NLP", "will": "", "explain": "samjhaenge", "it": "ise", "clearly": "spasht roop se"
}

def toy_translate(sentence):
    words = sentence.lower().replace(".", "").split()
    translated = [bilingual_dict.get(w, w) for w in words]
    return " ".join(w for w in translated if w)

print("EN:", sentence)
print("HI (toy, word-for-word):", toy_translate(sentence))
print("\nNote: real MT must reorder (Hindi is SOV) and handle morphology --")
print("this toy example deliberately shows why word-for-word MT fails, motivating Practical 8.")
```
